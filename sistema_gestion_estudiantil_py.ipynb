```python
class Estudiante:
    def __init__(self, codigo, nombre, programa, semestre, servicio):
        self.codigo = codigo
        self.nombre = nombre
        self.programa = programa
        self.semestre = semestre
        self.servicio = servicio

    def mostrar_informacion(self):
        return (
            f"Codigo: {self.codigo} | "
            f"Nombre: {self.nombre} | "
            f"Programa: {self.programa} | "
            f"Semestre: {self.semestre} | "
            f"Servicio: {self.servicio}"
        )


class NodoEstudiante:
    def __init__(self, estudiante):
        self.estudiante = estudiante
        self.siguiente = None


class ListaEstudiantes:
    def __init__(self):
        self.cabeza = None

    def insertar(self, estudiante):
        if self.buscar(estudiante.codigo) is not None:
            print("No se puede insertar: el codigo ya existe.")
            return False

        nuevo = NodoEstudiante(estudiante)

        if self.cabeza is None:
            self.cabeza = nuevo
        else:
            actual = self.cabeza

            while actual.siguiente is not None:
                actual = actual.siguiente

            actual.siguiente = nuevo

        print(f"Estudiante agregado: {estudiante.nombre}")
        return True

    def buscar(self, codigo):
        actual = self.cabeza

        while actual is not None:
            if actual.estudiante.codigo == codigo:
                return actual.estudiante

            actual = actual.siguiente

        return None

    def eliminar(self, codigo):
        if self.cabeza is None:
            print("La lista esta vacia.")
            return False

        if self.cabeza.estudiante.codigo == codigo:
            self.cabeza = self.cabeza.siguiente
            print("Estudiante eliminado correctamente.")
            return True

        anterior = self.cabeza
        actual = self.cabeza.siguiente

        while actual is not None:
            if actual.estudiante.codigo == codigo:
                anterior.siguiente = actual.siguiente
                print("Estudiante eliminado correctamente.")
                return True

            anterior = actual
            actual = actual.siguiente

        print("El codigo no existe.")
        return False

    def recorrer(self):
        if self.cabeza is None:
            print("La lista esta vacia.")
            return

        actual = self.cabeza

        while actual is not None:
            print(actual.estudiante.mostrar_informacion())
            actual = actual.siguiente


class NodoCola:
    def __init__(self, estudiante):
        self.estudiante = estudiante
        self.siguiente = None


class ColaLista:
    def __init__(self):
        self.frente = None
        self.final = None

    def esta_vacia(self):
        return self.frente is None

    def encolar(self, estudiante):
        nuevo = NodoCola(estudiante)

        if self.esta_vacia():
            self.frente = nuevo
            self.final = nuevo
        else:
            self.final.siguiente = nuevo
            self.final = nuevo

        print(f"Turno agregado: {estudiante.nombre}")
        return True

    def desencolar(self):
        if self.esta_vacia():
            print("La cola esta vacia. No hay estudiantes para atender.")
            return None

        estudiante = self.frente.estudiante
        self.frente = self.frente.siguiente

        if self.frente is None:
            self.final = None

        print(f"Atendiendo a: {estudiante.nombre}")
        return estudiante

    def mostrar(self):
        if self.esta_vacia():
            print("La cola esta vacia.")
            return

        actual = self.frente

        print("\nCOLA DE ATENCION")

        while actual is not None:
            print(
                f"{actual.estudiante.codigo} - "
                f"{actual.estudiante.nombre}"
            )
            actual = actual.siguiente


class ColaArreglo:
    def __init__(self, capacidad):
        self.capacidad = capacidad
        self.vector = [None] * capacidad
        self.frente = 0
        self.final = 0

    def esta_vacia(self):
        return self.frente == self.final

    def esta_llena(self):
        return self.final == self.capacidad

    def encolar(self, estudiante):
        if self.esta_llena():
            print("La cola esta llena. No se puede agregar otro estudiante.")
            return False

        self.vector[self.final] = estudiante
        self.final += 1

        print(f"Turno agregado: {estudiante.nombre}")
        return True

    def desencolar(self):
        if self.esta_vacia():
            print("La cola esta vacia. No hay estudiantes para atender.")
            return None

        estudiante = self.vector[self.frente]
        self.vector[self.frente] = None
        self.frente += 1

        print(f"Atendiendo a: {estudiante.nombre}")
        return estudiante

    def mostrar(self):
        if self.esta_vacia():
            print("La cola esta vacia.")
            return

        print("\nCOLA DE ATENCION")

        for i in range(self.frente, self.final):
            estudiante = self.vector[i]

            if estudiante is not None:
                print(
                    f"{estudiante.codigo} - "
                    f"{estudiante.nombre}"
                )


class NodoArbol:
    def __init__(self, estudiante):
        self.estudiante = estudiante
        self.izquierda = None
        self.derecha = None


class ArbolBinarioBusqueda:
    def __init__(self):
        self.raiz = None

    def insertar(self, estudiante):
        if self.buscar(estudiante.codigo) is not None:
            print("No se puede insertar: la clave ya existe.")
            return False

        if self.raiz is None:
            self.raiz = NodoArbol(estudiante)
            return True

        self.raiz = self._insertar_recursivo(
            self.raiz,
            estudiante
        )

        return True

    def _insertar_recursivo(self, nodo, estudiante):
        if estudiante.codigo < nodo.estudiante.codigo:

            if nodo.izquierda is None:
                nodo.izquierda = NodoArbol(estudiante)
            else:
                nodo.izquierda = self._insertar_recursivo(
                    nodo.izquierda,
                    estudiante
                )

        else:

            if nodo.derecha is None:
                nodo.derecha = NodoArbol(estudiante)
            else:
                nodo.derecha = self._insertar_recursivo(
                    nodo.derecha,
                    estudiante
                )

        return nodo

    def buscar(self, codigo):
        return self._buscar_recursivo(self.raiz, codigo)

    def _buscar_recursivo(self, nodo, codigo):
        if nodo is None:
            return None

        if codigo == nodo.estudiante.codigo:
            return nodo.estudiante

        if codigo < nodo.estudiante.codigo:
            return self._buscar_recursivo(
                nodo.izquierda,
                codigo
            )

        return self._buscar_recursivo(
            nodo.derecha,
            codigo
        )

    def preorden(self):
        resultado = []
        self._preorden_recursivo(self.raiz, resultado)
        return resultado

    def _preorden_recursivo(self, nodo, resultado):
        if nodo is not None:
            resultado.append(nodo.estudiante.codigo)
            self._preorden_recursivo(nodo.izquierda, resultado)
            self._preorden_recursivo(nodo.derecha, resultado)

    def inorden(self):
        resultado = []
        self._inorden_recursivo(self.raiz, resultado)
        return resultado

    def _inorden_recursivo(self, nodo, resultado):
        if nodo is not None:
            self._inorden_recursivo(nodo.izquierda, resultado)
            resultado.append(nodo.estudiante.codigo)
            self._inorden_recursivo(nodo.derecha, resultado)

    def postorden(self):
        resultado = []
        self._postorden_recursivo(self.raiz, resultado)
        return resultado

    def _postorden_recursivo(self, nodo, resultado):
        if nodo is not None:
            self._postorden_recursivo(nodo.izquierda, resultado)
            self._postorden_recursivo(nodo.derecha, resultado)
            resultado.append(nodo.estudiante.codigo)

    def altura(self):
        return self._altura_recursiva(self.raiz)

    def _altura_recursiva(self, nodo):
        if nodo is None:
            return 0

        izquierda = self._altura_recursiva(nodo.izquierda)
        derecha = self._altura_recursiva(nodo.derecha)

        return 1 + max(izquierda, derecha)


class NodoGrafo:
    def __init__(self, nombre):
        self.nombre = nombre
        self.vecinos = None
        self.siguiente = None


class Grafo:
    def __init__(self):
        self.nodos = None

    def agregar_nodo(self, nombre):
        if self.buscar_nodo(nombre) is not None:
            return False

        nuevo = NodoGrafo(nombre)
        nuevo.siguiente = self.nodos
        self.nodos = nuevo

        return True

    def buscar_nodo(self, nombre):
        actual = self.nodos

        while actual is not None:
            if actual.nombre == nombre:
                return actual

            actual = actual.siguiente

        return None

    def agregar_conexion(self, origen, destino):
        nodo_origen = self.buscar_nodo(origen)
        nodo_destino = self.buscar_nodo(destino)

        if nodo_origen is None or nodo_destino is None:
            print("Uno de los servicios no existe.")
            return False

        actual = nodo_origen.vecinos

        while actual is not None:
            if actual.nombre == destino:
                return False

            actual = actual.siguiente

        nuevo = NodoGrafo(destino)
        nuevo.siguiente = nodo_origen.vecinos
        nodo_origen.vecinos = nuevo

        return True

    def mostrar_vecinos(self, nombre):
        nodo = self.buscar_nodo(nombre)

        if nodo is None:
            print("El servicio no existe.")
            return

        print(f"\nVecinos de {nombre}:")

        if nodo.vecinos is None:
            print("No tiene conexiones.")
            return

        actual = nodo.vecinos

        while actual is not None:
            print(f"- {actual.nombre}")
            actual = actual.siguiente

    def estan_conectados(self, origen, destino):
        nodo = self.buscar_nodo(origen)

        if nodo is None:
            return False

        actual = nodo.vecinos

        while actual is not None:
            if actual.nombre == destino:
                return True

            actual = actual.siguiente

        return False


def crear_estudiantes():

    estudiantes = [
        Estudiante(50, "Laura Gomez", "Ingenieria de Software", 4, "Registro academico"),
        Estudiante(30, "Carlos Perez", "Administracion", 3, "Matriculas"),
        Estudiante(70, "Maria Rodriguez", "Contaduria", 5, "Tesoreria"),
        Estudiante(20, "Andres Lopez", "Ingenieria de Software", 2, "Biblioteca"),
        Estudiante(40, "Sofia Martinez", "Psicologia", 6, "Bienestar universitario"),
        Estudiante(60, "Daniel Torres", "Derecho", 4, "Admisiones"),
        Estudiante(80, "Valentina Castro", "Administracion", 2, "Matriculas"),
        Estudiante(10, "Juan Herrera", "Contaduria", 1, "Tesoreria"),
        Estudiante(25, "Camila Ruiz", "Psicologia", 3, "Bienestar universitario"),
        Estudiante(35, "Sebastian Diaz", "Derecho", 5, "Registro academico"),
        Estudiante(45, "Natalia Vargas", "Ingenieria de Software", 7, "Biblioteca"),
        Estudiante(55, "Felipe Moreno", "Administracion", 4, "Admisiones"),
        Estudiante(65, "Paula Jimenez", "Psicologia", 2, "Bienestar universitario"),
        Estudiante(75, "Mateo Ramirez", "Derecho", 6, "Tesoreria"),
        Estudiante(90, "Isabella Sanchez", "Ingenieria de Software", 8, "Registro academico")
    ]

    return estudiantes


def main():

    estudiantes = crear_estudiantes()

    print("\n==========================================")
    print("SISTEMA DE GESTION DE ATENCION ESTUDIANTIL")
    print("==========================================")

    print("\n--- LISTA ENLAZADA ---")

    lista = ListaEstudiantes()

    for estudiante in estudiantes[:5]:
        lista.insertar(estudiante)

    print("\nEstudiantes registrados:")
    lista.recorrer()

    print("\nBusqueda del codigo 30:")

    encontrado = lista.buscar(30)

    if encontrado is not None:
        print(encontrado.mostrar_informacion())
    else:
        print("Estudiante no encontrado.")

    print("\nEliminacion del codigo 30:")
    lista.eliminar(30)

    print("\nLista despues de eliminar:")
    lista.recorrer()

    print("\n--- COLA CON LISTA ENLAZADA ---")

    cola_lista = ColaLista()

    for estudiante in estudiantes[:4]:
        cola_lista.encolar(estudiante)

    cola_lista.mostrar()

    print("\nAtender estudiante:")
    cola_lista.desencolar()

    cola_lista.mostrar()

    print("\n--- COLA CON ARREGLO ---")

    cola_arreglo = ColaArreglo(4)

    for estudiante in estudiantes[:4]:
        cola_arreglo.encolar(estudiante)

    cola_arreglo.mostrar()

    print("\nIntentar agregar cuando esta llena:")
    cola_arreglo.encolar(estudiantes[4])

    print("\nAtender estudiante:")
    cola_arreglo.desencolar()

    cola_arreglo.mostrar()

    print("\n--- ARBOL BINARIO DE BUSQUEDA ---")

    arbol = ArbolBinarioBusqueda()

    for estudiante in estudiantes:
        arbol.insertar(estudiante)

    print("\nBusqueda del codigo 60:")

    encontrado = arbol.buscar(60)

    if encontrado is not None:
        print(encontrado.mostrar_informacion())
    else:
        print("Estudiante no encontrado.")

    print("\nPreorden:")
    print(arbol.preorden())

    print("\nInorden:")
    print(arbol.inorden())

    print("\nPostorden:")
    print(arbol.postorden())

    print("\nAltura del arbol:")
    print(arbol.altura())

    print("\nIntentar insertar una clave repetida:")
    arbol.insertar(estudiantes[0])

    print("\n--- EXPERIMENTO DE ALTURAS ---")

    arbol_desordenado = ArbolBinarioBusqueda()

    for estudiante in estudiantes:
        arbol_desordenado.insertar(estudiante)

    arbol_ordenado = ArbolBinarioBusqueda()

    estudiantes_ordenados = sorted(
        estudiantes,
        key=lambda estudiante: estudiante.codigo
    )

    for estudiante in estudiantes_ordenados:
        arbol_ordenado.insertar(estudiante)

    print("\nCantidad de datos: 15")
    print("Altura con datos desordenados:", arbol_desordenado.altura())
    print("Altura con datos ordenados:", arbol_ordenado.altura())

    print("\n--- GRAFO ---")

    grafo = Grafo()

    servicios = [
        "Registro academico",
        "Matriculas",
        "Bienestar universitario",
        "Biblioteca",
        "Tesoreria",
        "Admisiones"
    ]

    for servicio in servicios:
        grafo.agregar_nodo(servicio)

    grafo.agregar_conexion(
        "Registro academico",
        "Matriculas"
    )

    grafo.agregar_conexion(
        "Registro academico",
        "Biblioteca"
    )

    grafo.agregar_conexion(
        "Matriculas",
        "Tesoreria"
    )

    grafo.agregar_conexion(
        "Bienestar universitario",
        "Biblioteca"
    )

    grafo.agregar_conexion(
        "Admisiones",
        "Matriculas"
    )

    print("\nVecinos de Registro academico:")
    grafo.mostrar_vecinos("Registro academico")

    print("\nVerificar conexion:")

    if grafo.estan_conectados(
        "Registro academico",
        "Matriculas"
    ):
        print("Existe conexion directa.")
    else:
        print("No existe conexion directa.")

    print("\n--- CASOS LIMITE ---")

    lista_vacia = ListaEstudiantes()
    lista_vacia.eliminar(999)

    cola_vacia = ColaLista()
    cola_vacia.desencolar()

    estudiante_no_existente = arbol.buscar(999)

    if estudiante_no_existente is None:
        print("La clave 999 no existe en el arbol.")

    print("\n==========================================")
    print("FIN DEL PROGRAMA")
    print("==========================================")


if __name__ == "__main__":
    main()
```
